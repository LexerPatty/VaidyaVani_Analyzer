```python
import vertexai

import os
from dotenv import load_dotenv
from vertexai.generative_models import (
    GenerativeModel,
)


load_dotenv()

# WARNING .....
import warnings
warnings.filterwarnings("ignore")
```

```python
# CREDENTIALS PATH......................

credentials_path = os.getenv(
    "GOOGLE_APPLICATION_CREDENTIALS"
)
print(os.getenv("GOOGLE_APPLICATION_CREDENTIALS"))

absolute_path = os.path.abspath(
    credentials_path
)

os.environ[
    "GOOGLE_APPLICATION_CREDENTIALS"
] = absolute_path

PROJECT_ID = os.getenv("PROJECT_ID")

LOCATION = os.getenv(
    "LOCATION",
    "global"
)

vertexai.init(
    project=PROJECT_ID,
    location=LOCATION
)

print("Vertex AI Connected")

model = GenerativeModel(
    "gemini-2.5-flash"
)


```

```python
# QUESTION CODE ...........................
question = """
What is Artificial Intelligence?
Explain it in simple words with an example.
"""

try:
    # Generate response
    response = model.generate_content(question)

    print("=" * 80)
    print("QUESTION:")
    print(question)

    print("\n" + "=" * 80)
    print("ANSWER:")
    print(response.text)
    print("=" * 80)

except Exception as e:
    print("Error:")
    print(e)
```
